```python
import os
import re
from functools import lru_cache

import pandas as pd

REF_DIR = '/Users/roessner/Documents/PostDoc/Data/MixTCRviz/data_raw/CDR123/HomoSapiens'
OUT_ROOT = '.'
PRED_ROOT = 'ESMFold2_predictions'   # the {peptide} dirs live here

df = pd.read_csv('ttd_gil_ylq_nlv_glc_01-10-2026_TCRvdb.csv')


# ---------- germline reference ----------
def load_ref(segment):
    """{allele_name: (ungapped_full_seq, is_default_allele, is_functional)}"""
    ref = pd.read_csv(os.path.join(REF_DIR, f'{segment}_allele.csv'), index_col=0)
    out = {}
    for name, row in ref.iterrows():
        out[name] = (
            row['full'].replace('-', ''),
            str(name).endswith(str(row['default_allele'])),
            row['gene_type'] == 'F',
        )
    return out


REFS = {s: load_ref(s) for s in ['TRAV', 'TRAJ', 'TRBV', 'TRBJ']}


def _common_len(a, b):
    n = 0
    for x, y in zip(a, b):
        if x != y:
            break
        n += 1
    return n


@lru_cache(maxsize=None)
def assign_gene(seq, segment):
    """Best-matching germline gene for a full variable-domain sequence.

    Scored by longest common prefix (V) or longest common suffix (J): the V germline
    ends in a CDR3 stub and the J germline starts in one, both trimmed during
    recombination, so exact containment is not usable. Scoring runs per allele for
    discriminative power, then the allele suffix is dropped.
    """
    if not isinstance(seq, str) or not seq:
        return None
    is_j = segment.endswith('J')
    q = seq[::-1] if is_j else seq

    best_name, best_score = None, None
    for name, (germ, is_default, is_func) in REFS[segment].items():
        g = germ[::-1] if is_j else germ
        allele = int(name.split('*')[1]) if '*' in name else 99
        score = (_common_len(q, g), is_default, is_func, -allele)
        if best_score is None or score > best_score:
            best_name, best_score = name, score

    return best_name.split('*')[0]


def format_mhc(hla_long):
    """'A*02:01' -> 'HLA_A0201'"""
    return 'HLA_' + re.sub(r'[*:]', '', str(hla_long))


# ---------- infer V/J ----------
genes = pd.DataFrame({
    'TRAV': df['full_seq_alpha_aa'].map(lambda s: assign_gene(s, 'TRAV')),
    'TRAJ': df['full_seq_alpha_aa'].map(lambda s: assign_gene(s, 'TRAJ')),
    'TRBV': df['full_seq_beta_aa'].map(lambda s: assign_gene(s, 'TRBV')),
    'TRBJ': df['full_seq_beta_aa'].map(lambda s: assign_gene(s, 'TRBJ')),
}, index=df.index)


# ---------- write one {peptide}.csv per peptide ----------
COLS = ['id', 'cdr3_TRA', 'cdr3_TRB', 'TRAV', 'TRAJ', 'TRBV', 'TRBJ',
        'species', 'peptide', 'MHC_allele_a', 'tcr']

for peptide, sub in df.groupby('epitope_aa', sort=False):
    out_dir = os.path.join(OUT_ROOT, PRED_ROOT, peptide)
    os.makedirs(out_dir, exist_ok=True)

    out = pd.DataFrame({
        'id': [f'tcr_{i:04d}' for i in range(1, len(sub) + 1)],
        'cdr3_TRA': sub['cdr3_alpha_aa'].values,
        'cdr3_TRB': sub['cdr3_beta_aa'].values,
        'TRAV': genes.loc[sub.index, 'TRAV'].values,
        'TRAJ': genes.loc[sub.index, 'TRAJ'].values,
        'TRBV': genes.loc[sub.index, 'TRBV'].values,
        'TRBJ': genes.loc[sub.index, 'TRBJ'].values,
        'species': 'HomoSapiens',
        'peptide': peptide,
        'MHC_allele_a': [format_mhc(h) for h in sub['hla_long']],
        'tcr': sub['tcr_id'].values,
    })[COLS]

    out_path = os.path.join(out_dir, f'{peptide}.csv')
    out.to_csv(out_path, index=False)
    print(f'{peptide}: {len(out)} rows -> {out_path}')
```

```python
import os
import shutil

import pandas as pd

REF_DIR = '/Users/roessner/Documents/PostDoc/Data/MixTCRviz/data_raw/CDR123/HomoSapiens'
NEW_PEPTIDES = ['LTDEMIAQY', 'CINGVCWTV', 'LLWNGPMAV']
DROP_INCOMPLETE = True                              # single-chain rows can't be folded
GENE_ALIASES = {'TRBV6-2/6-3': 'TRBV6-2'}           # identical protein sequence

COLS = ['id', 'cdr3_TRA', 'cdr3_TRB', 'TRAV', 'TRAJ', 'TRBV', 'TRBJ',
        'species', 'peptide', 'MHC_allele_a']
CHAIN_COLS = ['cdr3_TRA', 'cdr3_TRB', 'TRAV', 'TRAJ', 'TRBV', 'TRBJ']

ref = {s: set(pd.read_csv(os.path.join(REF_DIR, f'{s}.csv'), index_col=0).index)
       for s in ['TRAV', 'TRAJ', 'TRBV', 'TRBJ']}

report = []
for peptide in NEW_PEPTIDES:
    src = f'ESMFold2_predictions/{peptide}/{peptide}.csv'

    # keep the untouched original once, so this stays re-runnable
    bak = f'ESMFold2_predictions/{peptide}/{peptide}_source.csv'
    if not os.path.exists(bak):
        shutil.copy2(src, bak)

    raw = pd.read_csv(bak)
    d = raw.copy()

    for s in ['TRAV', 'TRAJ', 'TRBV', 'TRBJ']:
        d[s] = d[s].replace(GENE_ALIASES)

    incomplete = d[CHAIN_COLS].isna().any(axis=1)
    if DROP_INCOMPLETE:
        d = d[~incomplete]

    d['MHC_allele_a'] = 'HLA_' + d['MHC'].astype(str)   # 'A0201' -> 'HLA_A0201'
    d['species'] = d['species'].fillna('HomoSapiens')
    d['peptide'] = d['peptide'].fillna(peptide)
    d = d.reset_index(drop=True)
    d['id'] = [f'tcr_{i:04d}' for i in range(1, len(d) + 1)]
    d = d[COLS]

    unknown = {s: sorted(set(d[s].dropna()) - ref[s]) for s in ref}
    unknown = {s: v for s, v in unknown.items() if v}

    d.to_csv(src, index=False)
    report.append(dict(peptide=peptide, n_raw=len(raw), dropped=int(incomplete.sum()),
                       n_out=len(d), mhc=d['MHC_allele_a'].iat[0],
                       unknown_genes=unknown or '-'))

display(pd.DataFrame(report))
```

# Compute AUC

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import roc_curve, roc_auc_score

peptides = ['TTDPSFLGRY', 'GILGFVFTL', 'YLQPRTFLL', 'NLVPMVATV', 'GLCTLVAML']
PRED_ROOTS = ['ESMFold2_predictions', 'ESMFold2_predictions_clariden']
SCORE = 'iptm_pair_mean'
DROP_AMBIGUOUS = True

# AUC reported in the paper (full AUC, not AUC01)
PAPER_AUC = {
    'TTDPSFLGRY': {'AF3 min-PAE': 0.71, 'tcrdist3': 0.83},
    'GILGFVFTL':  {'AF3 min-PAE': 0.68, 'tcrdist3': 0.87},
    'YLQPRTFLL':  {'AF3 min-PAE': 0.84, 'tcrdist3': 0.93},
    'NLVPMVATV':  {'AF3 min-PAE': 0.80, 'tcrdist3': 0.81},
    'GLCTLVAML':  {'AF3 min-PAE': 0.73, 'tcrdist3': 0.68},
}

meta = pd.read_csv('ttd_gil_ylq_nlv_glc_01-10-2026_TCRvdb.csv')
labels = meta.set_index('tcr_id')[['pos', 'neg', 'ambiguous']]
hla = meta.groupby('epitope_aa')['hla_long'].first()

for PRED_ROOT in PRED_ROOTS:
    fig, axes = plt.subplots(1, len(peptides), figsize=(15, 3.2), sharex=True, sharey=True)
    rows = []

    for cnt, peptide in enumerate(peptides):
        ax = axes[cnt]
        ax.plot([0, 1], [0, 1], ls='--', c='0.6', lw=0.8, zorder=0)   # random

        scores = (pd.read_csv(f'{PRED_ROOT}/{peptide}/output.csv')
                .groupby('ID', sort=False)[SCORE].max().rename('score'))
        df = (pd.read_csv(f'{PRED_ROOT}/{peptide}/{peptide}.csv').set_index('id')
            .join(scores, how='inner')
            .join(labels, on='tcr'))
        if DROP_AMBIGUOUS:
            df = df[df['ambiguous'] == 0]
        df = df.dropna(subset=['score', 'pos'])

        y_true, y_scores = df['pos'].values, df['score'].values
        fpr, tpr, _ = roc_curve(y_true, y_scores)
        auc01 = roc_auc_score(y_true, y_scores, max_fpr=0.1)
        auc = roc_auc_score(y_true, y_scores)
        rows.append(dict(peptide=peptide, n=len(df), n_pos=int(y_true.sum()),
                        n_neg=int((1 - y_true).sum()), auc=auc, auc01=auc01))

        ax.set_title(f'{peptide}, HLA-{hla[peptide]},\n{int(y_true.sum())}/{len(df)} TCRs',
                    fontsize='small')

        line, = ax.plot(fpr, tpr, label=f'{SCORE} (max of seeds)')
        ax.text(0.97, 0.30, f"AUC01 = {auc01:.2f} ({auc:.2f})",
                transform=ax.transAxes, ha='right', va='bottom',
                fontsize='x-small', color=line.get_color())
        for j, (method, val) in enumerate(PAPER_AUC[peptide].items()):
            ax.text(0.97, 0.23 - j * 0.07, f"{method} ({val:.2f})",
                    transform=ax.transAxes, ha='right', va='bottom',
                    fontsize='x-small', color='black')

        ax.set_xlabel('FPR')
        ax.set_ylabel('TPR')

    handles, labels_ = axes[-1].get_legend_handles_labels()
    axes[-1].legend(handles, labels_, fontsize='x-small', frameon=False,
                    bbox_to_anchor=(1.02, 1), loc='upper left')

    sns.despine()
    plt.tight_layout()
    plt.savefig(f'roc_{SCORE}.pdf', bbox_inches='tight')
    plt.show()

    res = pd.DataFrame(rows)
    display(res)
```

```python
import os
import glob
import pickle

import numpy as np
import pandas as pd
from Bio.PDB import MMCIFParser, PDBParser, Superimposer
from Bio.SeqUtils import seq1

PRED_ROOT = 'ESMFold2_predictions_clariden'
PAPER = '/Users/roessner/Documents/PostDoc/Data/260922_paper_YF_model'
GENE_REF = '/Users/roessner/Documents/PostDoc/Data/MixTCRviz/data_raw/CDR123/HomoSapiens'
EXP_DIR = f'{PAPER}/exp_structures/pdbs_mhc_align'
EXP_ANNO = f'{PAPER}/exp_structures/260305_data_structures_motif.csv'
EXP_REF_PDB = f'{EXP_DIR}/1AO7.pdb'
CACHE = f'{PRED_ROOT}/raw_cdr_traces.pkl'

N_MHC_CA = 180          # alpha1 + alpha2 platform
SCORE = 'iptm_pair_mean'

ungap = lambda s: str(s).replace('-', '').replace('.', '').upper()
V_REF = {'TRA': pd.read_csv(f'{GENE_REF}/TRAV.csv', index_col=0),
         'TRB': pd.read_csv(f'{GENE_REF}/TRBV.csv', index_col=0)}

# the exp structures are already in one common MHC frame, so one of them defines it
ref_chain = PDBParser(QUIET=True).get_structure('ref', EXP_REF_PDB)[0]['A']
REF_CA = [r['CA'] for r in ref_chain if r.id[0] == ' ' and 'CA' in r][:N_MHC_CA]
mmcif, pdbp = MMCIFParser(QUIET=True), PDBParser(QUIET=True)


def traces_model(cif_path, cdrs):
    """Superpose on MHC, return the raw (unresampled) CA trace of each CDR."""
    st = mmcif.get_structure('m', cif_path)[0]
    mob = [r['CA'] for r in st['MHC'] if r.id[0] == ' ' and 'CA' in r][:N_MHC_CA]
    if len(mob) != N_MHC_CA:
        return None, 'MHC too short'
    sup = Superimposer()
    sup.set_atoms(REF_CA, mob)
    sup.apply(list(st.get_atoms()))

    out = {}
    for arm, chain in [('TRA', 'TCRA'), ('TRB', 'TCRB')]:
        res = [r for r in st[chain] if r.id[0] == ' ' and 'CA' in r]
        seq = ''.join(seq1(r.get_resname()) for r in res)
        out[arm] = {}
        for i, cd in enumerate(cdrs[arm], 1):
            j = seq.find(cd)
            if j < 0:
                return None, f'{arm} cdr{i} not found'
            out[arm][f'cdr{i}'] = np.array([r['CA'].get_coord() for r in res[j:j + len(cd)]])
    return out, sup.rms


def traces_exp(pdb_path, arm_cdrs):
    """Chain ids vary here, so detect each arm as the chain containing its three CDRs."""
    st = pdbp.get_structure('x', pdb_path)
    chains = {}
    for ch in st[0]:
        res = [r for r in ch if r.id[0] == ' ' and 'CA' in r]
        if res:
            chains[ch.id] = (res, ''.join(seq1(r.get_resname()) for r in res))
    out = {}
    for arm, cdrs in arm_cdrs.items():
        hit = next(((res, seq) for res, seq in chains.values()
                    if all(c in seq for c in cdrs)), None)
        if hit is None:
            return None
        res, seq = hit
        out[arm] = {}
        for i, cd in enumerate(cdrs, 1):
            j = seq.find(cd)
            out[arm][f'cdr{i}'] = np.array([r['CA'].get_coord() for r in res[j:j + len(cd)]])
    return out


# ---------- experimental ----------
anno = pd.read_csv(EXP_ANNO)
anno = anno[anno['MHC'] == 'classI'].copy()
arm_cols = {'TRA': ['cdr1_TRA', 'cdr2_TRA', 'cdr3_TRA'],
            'TRB': ['cdr1_TRB', 'cdr2_TRB', 'cdr3_TRB']}

exp_traces, exp_meta = {}, []
for _, row in anno.iterrows():
    pid = str(row['PDB']).upper()
    f = os.path.join(EXP_DIR, f'{pid}.pdb')
    if not os.path.exists(f):
        continue
    cdrs = {a: [str(row[c]).upper() for c in cols] for a, cols in arm_cols.items()}
    if any(x == 'NAN' for v in cdrs.values() for x in v):
        continue
    t = traces_exp(f, cdrs)
    if t is not None:
        exp_traces[pid] = t
        exp_meta.append({'PDB': pid, 'Epitope': row.get('Epitope'),
                         'MHC Name': row.get('MHC Name')})
print(f'experimental: {len(exp_traces)} / {len(anno)} class-I structures')

# ---------- models ----------
mod_traces, mod_meta, skipped = {}, [], []
for path in sorted(glob.glob(f'{PRED_ROOT}/*/output.csv')):
    peptide = os.path.basename(os.path.dirname(path))
    tbl_path = f'{PRED_ROOT}/{peptide}/{peptide}.csv'
    if not os.path.exists(tbl_path):            # skips bck.* dirs
        continue
    ranked = pd.read_csv(path).sort_values(SCORE, ascending=False)
    tbl = pd.read_csv(tbl_path).set_index('id')

    n_ok = 0
    for tid, r in tbl.iterrows():
        cand = ranked[ranked['ID'] == tid]
        if cand.empty:
            skipped.append((peptide, tid, 'no fold')); continue

        cif = b = None                           # best-scoring seed present on disk
        for _, c in cand.iterrows():
            p = f'{PRED_ROOT}/{peptide}/models/{tid}_seed{int(c.seed)}_sample{int(c.sample_idx)}.cif'
            if os.path.exists(p):
                cif, b = p, c
                break
        if cif is None:
            skipped.append((peptide, tid, 'no cif')); continue

        try:   # CDR1/2 are germline, so they come from the V gene reference
            cdrs = {'TRA': [ungap(V_REF['TRA'].loc[r.TRAV, 'CDR1']),
                            ungap(V_REF['TRA'].loc[r.TRAV, 'CDR2']), r.cdr3_TRA],
                    'TRB': [ungap(V_REF['TRB'].loc[r.TRBV, 'CDR1']),
                            ungap(V_REF['TRB'].loc[r.TRBV, 'CDR2']), r.cdr3_TRB]}
        except KeyError as e:
            skipped.append((peptide, tid, f'gene {e}')); continue

        t, info = traces_model(cif, cdrs)
        if t is None:
            skipped.append((peptide, tid, info)); continue
        mod_traces[(peptide, tid)] = t
        mod_meta.append({'peptide': peptide, 'id': tid, 'tcr': r.tcr,
                         'seed': int(b.seed), 'score': b[SCORE], 'mhc_rmsd': info})
        n_ok += 1
    print(f'{peptide}: {n_ok}/{len(tbl)}')

with open(CACHE, 'wb') as fh:
    pickle.dump({'exp_traces': exp_traces, 'exp_meta': pd.DataFrame(exp_meta),
                 'mod_traces': mod_traces, 'mod_meta': pd.DataFrame(mod_meta)}, fh)

mm = pd.DataFrame(mod_meta)
print(f'\n{len(mod_traces)} model + {len(exp_traces)} exp traces -> {CACHE}')
print('MHC RMSD: median %.2f  max %.2f' % (mm.mhc_rmsd.median(), mm.mhc_rmsd.max()))
if skipped:
    display(pd.DataFrame(skipped, columns=['peptide', 'id', 'why'])['why'].value_counts())
```

```python
import pickle

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.colors import LinearSegmentedColormap
from matplotlib.lines import Line2D
from matplotlib.patches import Patch
from sklearn.decomposition import PCA

PRED_ROOT = 'ESMFold2_predictions_clariden'
CACHE = f'{PRED_ROOT}/raw_cdr_traces.pkl'
MIN_SCORE = 0.0            # paper used 0.8
EXCLUDE_EXP = ['9C3E']     # reverse-polarity TCR
NCOL = 4
CDRS = ['cdr1', 'cdr2', 'cdr3']

kde_cmap = LinearSegmentedColormap.from_list('paper_grey',
                                             ['#E4E8EB', '#C4CACF', '#4B535A'])

with open(CACHE, 'rb') as fh:
    D = pickle.load(fh)
exp_traces, mod_traces = D['exp_traces'], D['mod_traces']
mod_meta = D['mod_meta'].set_index(['peptide', 'id'])
exp_keys = [k for k in exp_traces if k not in EXCLUDE_EXP]


def resample_loop(coords, n):
    coords = np.asarray(coords, float)
    if len(coords) == n:
        return coords
    if len(coords) == 1:
        return np.repeat(coords, n, axis=0)
    d = np.r_[0, np.cumsum(np.linalg.norm(np.diff(coords, axis=0), axis=1))]
    if d[-1] == 0:
        return np.repeat(coords[:1], n, axis=0)
    u = d / d[-1]
    return np.column_stack([np.interp(np.linspace(0, 1, n), u, coords[:, k])
                            for k in range(3)])


def vectorise(traces, keys, targets):
    X = []
    for k in keys:
        t, vec = traces[k], []
        for arm in ['TRA', 'TRB']:
            for c in CDRS:
                vec.extend(resample_loop(t[arm][c], targets[arm][c]).flatten())
        X.append(vec)
    return np.asarray(X)


peptides = sorted({p for p, _ in mod_traces})
nrow = int(np.ceil(len(peptides) / NCOL))
fig, axes = plt.subplots(nrow, NCOL, figsize=(3.1 * NCOL, 3.0 * nrow))
axes = np.atleast_1d(axes).ravel()
report = []

for ax, peptide in zip(axes, peptides):
    keys = [k for k in mod_traces if k[0] == peptide
            and mod_meta.loc[k, 'score'] >= MIN_SCORE]

    # points per loop = MEDIAN loop length across THIS epitope's models,
    # then the experimental structures are resampled to the same targets
    targets = {arm: {c: int(np.round(np.median([len(mod_traces[k][arm][c]) for k in keys])))
                     for c in CDRS} for arm in ['TRA', 'TRB']}

    Xexp = vectorise(exp_traces, exp_keys, targets)
    Xmod = vectorise(mod_traces, keys, targets)

    pca = PCA(n_components=2, random_state=0).fit(Xexp)
    pe, pm = pca.transform(Xexp), pca.transform(Xmod)
    dfe = pd.DataFrame(pe, columns=['PC1', 'PC2'])

    sns.kdeplot(data=dfe, x='PC1', y='PC2', ax=ax, fill=True,
                thresh=0.05, levels=8, cmap=kde_cmap, alpha=0.9)
    sns.kdeplot(data=dfe, x='PC1', y='PC2', ax=ax, fill=False,
                thresh=0.05, levels=8, color='#8A9299', linewidths=0.4, alpha=0.8)
    ax.scatter(pm[:, 0], pm[:, 1], marker='x', c='black', s=12, linewidths=0.5)

    lo, hi = np.quantile(np.r_[pe, pm], [0.005, 0.995], axis=0)
    ax.set_xlim(lo[0] - 8, hi[0] + 8)
    ax.set_ylim(lo[1] - 8, hi[1] + 8)

    t = targets
    ax.set_title(f"{peptide}  (n={len(keys)})\n"
                 f"a {t['TRA']['cdr1']}/{t['TRA']['cdr2']}/{t['TRA']['cdr3']}  "
                 f"b {t['TRB']['cdr1']}/{t['TRB']['cdr2']}/{t['TRB']['cdr3']}",
                 fontsize='small')
    ax.set_xlabel('PC1'); ax.set_ylabel('PC2')
    report.append(dict(peptide=peptide, n=len(keys), n_features=Xexp.shape[1],
                       **{f'{a}_{c}': targets[a][c] for a in ['TRA', 'TRB'] for c in CDRS},
                       var_PC1=pca.explained_variance_ratio_[0],
                       var_PC2=pca.explained_variance_ratio_[1]))

for ax in axes[len(peptides):]:
    ax.axis('off')

h = [Patch(facecolor='#C4CACF', edgecolor='#8A9299', lw=0.4, label='Experimental structures'),
     Line2D([0], [0], marker='x', color='black', ls='', ms=5, mew=1.2,
            label='ESMFold2 predictions')]
axes[0].legend(handles=h, frameon=False, loc='upper left', fontsize=7,
               handlelength=1, handletextpad=0.5)

sns.despine()
plt.tight_layout()
plt.show()

display(pd.DataFrame(report).round(3))
```

```python
import os
import glob
import pickle

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from Bio.PDB import MMCIFParser, PDBParser, Superimposer

PRED_ROOT = 'ESMFold2_predictions_clariden'
PAPER = '/Users/roessner/Documents/PostDoc/Data/260922_paper_YF_model'
EXP_REF_PDB = f'{PAPER}/exp_structures/pdbs_mhc_align/1AO7.pdb'
CACHE = f'{PRED_ROOT}/peptide_ca.pkl'
N_MHC_CA = 180
SCORE = 'iptm_pair_mean'

# ---- extract peptide CA in the common MHC frame (cached; ~2 min the first time) ----
if not os.path.exists(CACHE):
    ref = [r['CA'] for r in PDBParser(QUIET=True).get_structure('r', EXP_REF_PDB)[0]['A']
           if r.id[0] == ' ' and 'CA' in r][:N_MHC_CA]
    parser = MMCIFParser(QUIET=True)
    pep = {}
    for path in sorted(glob.glob(f'{PRED_ROOT}/*/output.csv')):
        peptide = os.path.basename(os.path.dirname(path))
        tbl_path = f'{PRED_ROOT}/{peptide}/{peptide}.csv'
        if not os.path.exists(tbl_path):
            continue
        ranked = pd.read_csv(path).sort_values(SCORE, ascending=False)
        coords = []
        for tid in pd.read_csv(tbl_path)['id']:
            f = None
            for _, c in ranked[ranked['ID'] == tid].iterrows():
                q = (f'{PRED_ROOT}/{peptide}/models/'
                     f'{tid}_seed{int(c.seed)}_sample{int(c.sample_idx)}.cif')
                if os.path.exists(q):
                    f = q
                    break
            if f is None:
                continue
            st = parser.get_structure('m', f)[0]
            mob = [r['CA'] for r in st['MHC'] if r.id[0] == ' ' and 'CA' in r][:N_MHC_CA]
            if len(mob) != N_MHC_CA:
                continue
            sup = Superimposer()
            sup.set_atoms(ref, mob)
            sup.apply(list(st.get_atoms()))
            # NOTE: no peptide-on-peptide superposition, so RMSF keeps the
            # placement/register variation in the groove, not just loop shape
            coords.append(np.array([r['CA'].get_coord() for r in st['PEPTIDE']
                                    if r.id[0] == ' ' and 'CA' in r]))
        L = min(len(c) for c in coords)
        pep[peptide] = np.array([c[:L] for c in coords])
        print(f'{peptide}: {len(coords)} models, length {L}')
    with open(CACHE, 'wb') as fh:
        pickle.dump(pep, fh)
else:
    with open(CACHE, 'rb') as fh:
        pep = pickle.load(fh)

# ---- per-residue RMSF ----
rmsf = {p: np.sqrt(((A - A.mean(0)) ** 2).sum(2).mean(0)) for p, A in pep.items()}

order = sorted(rmsf, key=lambda p: np.sqrt((rmsf[p] ** 2).mean()))
pal = dict(zip(order, sns.color_palette('colorblind', len(order))))

fig, ax = plt.subplots(figsize=(6, 3.6))
for peptide in order:
    r = rmsf[peptide]
    ax.plot(np.arange(1, len(r) + 1), r, '-o', ms=3.5, lw=1.4,
            color=pal[peptide], label=f'{peptide}  (n={len(pep[peptide])})')

ax.set_xticks(range(1, max(len(r) for r in rmsf.values()) + 1))
ax.set_xlabel('Peptide position')
ax.set_ylabel(r'C$\alpha$ RMSF ($\AA$)')
ax.legend(frameon=False, fontsize=7, bbox_to_anchor=(1.02, 1), loc='upper left')
sns.despine()
plt.tight_layout()
plt.show()

display(pd.DataFrame({'n': {p: len(pep[p]) for p in order},
                      'L': {p: len(rmsf[p]) for p in order},
                      'RMSF': {p: np.sqrt((rmsf[p] ** 2).mean()) for p in order},
                      'anchors': {p: np.mean([rmsf[p][1], rmsf[p][-1]]) for p in order},
                      'max': {p: rmsf[p].max() for p in order},
                      'at_pos': {p: int(np.argmax(rmsf[p]) + 1) for p in order}}).round(2))
```

```python
import os
import pickle

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.lines import Line2D
from Bio.PDB import MMCIFParser, PDBParser, Superimposer

RUNS = {'no template': 'ESMFold2_predictions_clariden',
        'pMHC template': 'ESMFold2_predictions_clariden_template'}
PAPER = '/Users/roessner/Documents/PostDoc/Data/260922_paper_YF_model'
EXP_REF_PDB = f'{PAPER}/exp_structures/pdbs_mhc_align/1AO7.pdb'
N_MHC_CA = 180
DROP_AMBIGUOUS = True
COLORS = {'no template': '#4C72B0', 'pMHC template': '#DD8452'}
STYLE = {'binder': dict(ls='-', mfc=None), 'decoy': dict(ls=(0, (4, 2)), mfc='white')}

ref = [r['CA'] for r in PDBParser(QUIET=True).get_structure('r', EXP_REF_PDB)[0]['A']
       if r.id[0] == ' ' and 'CA' in r][:N_MHC_CA]
parser = MMCIFParser(QUIET=True)


def peptide_ca(root, peptide):
    """{tcr_id: {seed: (L,3) peptide CA}} in the common MHC frame. Cached per run."""
    cache = f'{root}/{peptide}/peptide_ca_allseeds.pkl'
    if os.path.exists(cache):
        return pickle.load(open(cache, 'rb'))
    out = {}
    for _, c in pd.read_csv(f'{root}/{peptide}/output.csv').iterrows():
        f = f'{root}/{peptide}/models/{c.ID}_seed{int(c.seed)}_sample{int(c.sample_idx)}.cif'
        if not os.path.exists(f):
            continue
        st = parser.get_structure('m', f)[0]
        mob = [r['CA'] for r in st['MHC'] if r.id[0] == ' ' and 'CA' in r][:N_MHC_CA]
        if len(mob) != N_MHC_CA:
            continue
        sup = Superimposer()
        sup.set_atoms(ref, mob)
        sup.apply(list(st.get_atoms()))
        out.setdefault(c.ID, {})[int(c.seed)] = np.array(
            [r['CA'].get_coord() for r in st['PEPTIDE'] if r.id[0] == ' ' and 'CA' in r])
    pickle.dump(out, open(cache, 'wb'))
    print(f'  cached {root}/{peptide} ({len(out)} TCRs)')
    return out


base, tmpl = RUNS['no template'], RUNS['pMHC template']
peptides = sorted(p for p in os.listdir(tmpl)              # only epitopes run both ways
                  if os.path.exists(f'{tmpl}/{p}/output.csv')
                  and os.path.exists(f'{base}/{p}/output.csv'))
print('epitopes:', peptides)

meta = pd.read_csv('ttd_gil_ylq_nlv_glc_01-10-2026_TCRvdb.csv')
labels = meta.set_index('tcr_id')[['pos', 'ambiguous']]

prof, summ, labelled = {}, [], {}
for pep in peptides:
    for run, root in RUNS.items():
        d = peptide_ca(root, pep)
        tbl = pd.read_csv(f'{root}/{pep}/{pep}.csv').set_index('id')
        ids = [t for t, v in d.items() if len(v) == 3]
        L = min(len(x) for t in ids for x in d[t].values())

        A = np.array([np.mean([d[t][s][:L] for s in sorted(d[t])], axis=0) for t in ids])
        prof[(pep, run, 'all')] = np.sqrt(((A - A.mean(0)) ** 2).sum(2).mean(0))

        info = pd.DataFrame({'id': ids}).join(tbl['tcr'], on='id').join(labels, on='tcr')
        labelled[pep] = info['pos'].notna().any()
        if labelled[pep]:
            if DROP_AMBIGUOUS:
                info = info[info['ambiguous'] == 0]
            pos = {i: k for k, i in enumerate(ids)}
            for lb, v in [('binder', 1), ('decoy', 0)]:
                sub = A[[pos[i] for i in info.loc[info['pos'] == v, 'id']]]
                prof[(pep, run, lb)] = np.sqrt(((sub - sub.mean(0)) ** 2).sum(2).mean(0))

        # within-TCR spread across its own 3 seeds = pure model noise
        W = np.array([np.sqrt(((np.stack([d[t][s][:L] for s in sorted(d[t])])
                                - np.mean([d[t][s][:L] for s in sorted(d[t])], axis=0))
                               ** 2).sum(2).mean(0)) for t in ids])
        prof[(pep, run, 'within-seed')] = W.mean(0)

        g = lambda v: np.sqrt((v ** 2).mean())
        row = dict(peptide=pep, run=run, n=len(ids),
                   RMSF_all=g(prof[(pep, run, 'all')]), within_seed=g(W.mean(0)),
                   anchors=np.mean([prof[(pep, run, 'all')][1], prof[(pep, run, 'all')][-1]]))
        if labelled[pep]:
            row['RMSF_binder'] = g(prof[(pep, run, 'binder')])
            row['RMSF_decoy'] = g(prof[(pep, run, 'decoy')])
            row['ratio_b_d'] = row['RMSF_binder'] / row['RMSF_decoy']
        summ.append(row)

n = len(peptides)
fig, axes = plt.subplots(3, n, figsize=(2.9 * n, 8.6), sharey='row')
for j, pep in enumerate(peptides):
    for i, (key, title) in enumerate([('all', 'between-TCR'), (None, 'by label'),
                                      ('within-seed', 'within-TCR (seeds)')]):
        ax = axes[i, j]
        if key is not None:
            for run in RUNS:
                r = prof[(pep, run, key)]
                ax.plot(np.arange(1, len(r) + 1), r, '-o', ms=3.5, lw=1.4,
                        color=COLORS[run], label=run)
        elif labelled[pep]:
            for run in RUNS:
                for lb in ['binder', 'decoy']:
                    r, st = prof[(pep, run, lb)], STYLE[lb]
                    ax.plot(np.arange(1, len(r) + 1), r, ls=st['ls'], marker='o',
                            ms=3.5, lw=1.4, color=COLORS[run],
                            mfc=st['mfc'] or COLORS[run], mew=1.1)
        else:
            ax.text(0.5, 0.5, 'no labels', ha='center', va='center',
                    transform=ax.transAxes, fontsize=8, color='0.5')
        ax.set_title(f'{pep}\n{title}' if i == 0 else title,
                     fontsize='small' if i == 0 else 7)
        ax.set_xlabel('Peptide position', fontsize=8)
        if j == 0:
            ax.set_ylabel(r'C$\alpha$ RMSF ($\AA$)')

axes[0, 0].legend(frameon=False, fontsize=6.5)
# colour means run, line style means label (two legends, on the first labelled panel)
k = next(j for j, p in enumerate(peptides) if labelled[p])
h_run = [Line2D([], [], color=COLORS[r], lw=1.4, label=r) for r in RUNS]
h_lab = [Line2D([], [], color='0.3', lw=1.4, ls=STYLE[l]['ls'], marker='o', ms=3.5,
                mfc=STYLE[l]['mfc'] or '0.3', mew=1.1, label=l) for l in STYLE]
leg = axes[1, k].legend(handles=h_run, frameon=False, fontsize=6, loc='upper left',
                        handlelength=2.4)
axes[1, k].add_artist(leg)
axes[1, k].legend(handles=h_lab, frameon=False, fontsize=6, loc='upper right',
                  handlelength=2.4)

fig.suptitle('Peptide RMSF: pMHC prompt vs none', fontsize='small')
sns.despine()
plt.tight_layout()
plt.show()

display(pd.DataFrame(summ).set_index(['peptide', 'run']).round(3))
```

```python
import os

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.lines import Line2D

RUNS = {'Fast-Opt': 'ESMFold2_predictions_clariden_optimized',
        'Full-Opt': 'ESMFold2_predictions_clariden_full_checkpoint_optimized'}
SCORE = 'iptm_pair_mean'
DROP_AMBIGUOUS = True
NCOL = 4

# multipliers on each epitope's BINDER COUNT. strings, so the filenames match the
# R side exactly (R turns 1.0 into "1", Python's f'{1.0}' gives "1.0")
FRACS = ['0.5', '0.8', '1.0', '1.2']
OUT_DIR = 'TEMPO'
WRITE_CSV = True
TEMPO_COLS = ['id', 'cdr3_TRA', 'cdr3_TRB', 'TRAV', 'TRAJ', 'TRBV', 'TRBJ',
              'species', 'peptide', 'MHC_allele_a', 'tcr']

COLORS = {'binder': '#1f77b4', 'decoy': '#ff7f0e', 'unlabelled': '#9e9e9e'}
# purples: distinct from the binder blue and decoy orange
FRACCOL = dict(zip(FRACS, sns.color_palette('Purples', len(FRACS) + 2)[2:]))
NAMES = list(RUNS)
REF = RUNS[NAMES[0]]

# expected binders for the unlabelled epitopes (Messemaker Fig E: binders / total)
EXPECTED = {'CINGVCWTV': (174, 228), 'LTDEMIAQY': (75, 127), 'LLWNGPMAV': (162, 215)}

meta = pd.read_csv('ttd_gil_ylq_nlv_glc_01-10-2026_TCRvdb.csv')
labels = meta.set_index('tcr_id')[['pos', 'ambiguous']]


def hla_of(pep):
    a = str(pd.read_csv(f'{REF}/{pep}/{pep}.csv')['MHC_allele_a'].iloc[0]).replace('HLA_', '')
    return f'HLA-{a[0]}*{a[1:3]}:{a[3:]}'


avail = {k: {p for p in os.listdir(r) if os.path.exists(f'{r}/{p}/output.csv')}
         for k, r in RUNS.items()}
peptides = sorted(set.intersection(*avail.values()))
for p in sorted(set.union(*avail.values()) - set(peptides)):
    print(f'EXCLUDED {p}: missing in ' + ', '.join(k for k in NAMES if p not in avail[k]))

frames = {}
for pep in peptides:
    cols = {r: (pd.read_csv(f'{root}/{pep}/output.csv')
                .groupby('ID', sort=False)[SCORE].max().rename(r))
            for r, root in RUNS.items()}
    m = pd.concat(cols.values(), axis=1, join='inner')
    tbl = pd.read_csv(f'{REF}/{pep}/{pep}.csv').set_index('id')
    d = tbl.join(m, how='inner').reset_index().set_index('tcr').join(labels)
    d = d.dropna(subset=NAMES)
    if DROP_AMBIGUOUS and d['pos'].notna().any():
        d = d[d['ambiguous'] == 0]
    d['score'] = d[NAMES].max(axis=1)                 # best complex across both models
    d['best_model'] = d[NAMES].idxmax(axis=1)
    d['label'] = (d['pos'].map({1: 'binder', 0: 'decoy'})
                  if d['pos'].notna().any() else 'unlabelled')
    frames[pep] = d

if WRITE_CSV:
    os.makedirs(OUT_DIR, exist_ok=True)

nrow = int(np.ceil(len(peptides) / NCOL))
fig, axes = plt.subplots(nrow, NCOL, figsize=(3.3 * NCOL, 3.2 * nrow),
                         sharex=True, sharey=True)
axes = np.atleast_1d(axes).ravel()
rows = []
for ax, pep in zip(axes, peptides):
    d = frames[pep]
    lab = d['label'].iloc[0] != 'unlabelled'
    n = len(d)

    for g in (['decoy', 'binder'] if lab else ['unlabelled']):
        v = d.loc[d['label'] == g, 'score']
        sns.kdeplot(x=v, ax=ax, fill=True, alpha=0.35, lw=1.6, color=COLORS[g],
                    cut=0, label=f'{g} (n={len(v)})')
        ax.axvline(v.median(), color=COLORS[g], ls=':', lw=1.2)

    if lab:
        n_bind, src = int((d['pos'] == 1).sum()), 'observed binders'
    else:
        kb, nt = EXPECTED[pep]
        n_bind, src = int(round(kb / nt * n)), f'expected {kb}/{nt}'

    top = ax.get_ylim()[1]
    for frac in FRACS:
        k = int(np.clip(round(float(frac) * n_bind), 1, n))
        thr = np.sort(d['score'].to_numpy())[::-1][k - 1]      # k-th largest score
        sel = d.nlargest(k, 'score').copy()

        ax.axvline(thr, color=FRACCOL[frac], ls='--', lw=1.3)
        ax.annotate(str(k), xy=(thr, top * 0.98), xytext=(2, 0),
                    textcoords='offset points', fontsize=5.5, rotation=90,
                    va='top', ha='left', color=FRACCOL[frac])

        r = dict(peptide=pep, frac=frac, n=n, n_binders=n_bind, k=k,
                 k_source=src, threshold=thr)
        if lab:
            r['precision_at_k'] = (sel['pos'] == 1).mean()
            r['recall_at_k'] = (sel['pos'] == 1).sum() / (d['pos'] == 1).sum()
        if WRITE_CSV:
            out = sel.reset_index()                   # 'tcr' is the index, bring it back
            missing = [c for c in TEMPO_COLS if c not in out.columns]
            assert not missing, f'{pep}: missing columns {missing}'
            path = f'{OUT_DIR}/{pep}_{frac}.csv'
            out[TEMPO_COLS].to_csv(path, index=False)
            r['csv'] = os.path.basename(path)
        rows.append(r)

    cnt = f"{int((d['pos'] == 1).sum())}/{n} TCRs" if lab else f'{n} TCRs, no labels'
    ax.set_title(f'{pep}, {hla_of(pep)},\n{cnt}', fontsize='small')
    ax.legend(frameon=False, fontsize='x-small', loc='upper left')
    ax.set_xlabel(f'max({SCORE}) across both optimized models', fontsize=8)
    ax.set_ylabel('Density')
for ax in axes[len(peptides):]:
    ax.axis('off')

h = [Line2D([], [], color=FRACCOL[f], ls='--', lw=1.3, label=f) for f in FRACS]
sns.despine()
plt.tight_layout(rect=[0, 0.05, 1, 1])
fig.legend(handles=h, title='FRAC_OF_BINDERS', frameon=False, fontsize=7.5,
           title_fontsize=7.5, loc='lower center', ncol=len(FRACS) + 1,
           bbox_to_anchor=(0.5, 0.0))
plt.show()

display(pd.DataFrame(rows).round(3))
```

```python
import os

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import rankdata
from sklearn.metrics import roc_auc_score, roc_curve

FRACS = ['0.5', '0.8', '1.0', '1.2']
ALL_FRAC = 'all'            # TEMPO trained on every TCR, the no-selection control
COMBO_FRAC = '1.0'          # which TEMPO model goes into the combination
W = 0.5                     # weight on TEMPO; 1-W on iptm
EPITOPES = ['TTDPSFLGRY', 'GILGFVFTL', 'YLQPRTFLL', 'NLVPMVATV', 'GLCTLVAML']
OUT_ROOT = 'TEMPO_out'
DROP_AMBIGUOUS = True

REF_RUNS = {'Fast-Opt': 'ESMFold2_predictions_clariden_optimized',
            'Full-Opt': 'ESMFold2_predictions_clariden_full_checkpoint_optimized'}
REF_SCORE = 'iptm_pair_mean'
REF_LABEL = 'iptm max(Fast-Opt, Full-Opt)'
ALL_LABEL = 'TEMPO all (no selection)'
COMBO_LABEL = f'{W:g}*TEMPO {COMBO_FRAC} + {1-W:g}*iptm (ranks)'

meta = pd.read_csv('ttd_gil_ylq_nlv_glc_01-10-2026_TCRvdb.csv')
labels = meta.set_index('tcr_id')[['pos', 'ambiguous']]
pal = dict(zip(FRACS, sns.color_palette('viridis', len(FRACS))))
R = lambda v: rankdata(v) / len(v)          # rank-normalise: the two scales differ


def reference(ep):
    cols = {}
    for run, root in REF_RUNS.items():
        if not os.path.exists(f'{root}/{ep}/output.csv'):
            return None
        sc = (pd.read_csv(f'{root}/{ep}/output.csv')
              .groupby('ID', sort=False)[REF_SCORE].max().rename(run))
        tbl = pd.read_csv(f'{root}/{ep}/{ep}.csv').set_index('id')
        cols[run] = tbl.join(sc, how='inner').set_index('tcr')[run]
    d = pd.concat(cols.values(), axis=1, join='inner').join(labels)
    if DROP_AMBIGUOUS:
        d = d[d['ambiguous'] == 0]
    d = d.dropna(subset=list(REF_RUNS) + ['pos'])
    d['iptm'] = d[list(REF_RUNS)].max(axis=1)
    return d


def tempo(ep, frac):
    p = f'{OUT_ROOT}/{ep}_{frac}/all_tcrs.csv'
    if not os.path.exists(p):
        print(f'missing: {p}')
        return None
    return pd.read_csv(p).set_index('tcr')['score'].rename('tempo')


def scored(t):
    """join labels to a TEMPO score series and drop what cannot be evaluated"""
    d = t.to_frame().join(labels)
    if DROP_AMBIGUOUS:
        d = d[d['ambiguous'] == 0]
    return d.dropna(subset=['tempo', 'pos'])


res, sweep = [], []
fig, axes = plt.subplots(1, len(EPITOPES), figsize=(3.3 * len(EPITOPES), 3.5),
                         sharex=True, sharey=True)
for ax, ep in zip(np.atleast_1d(axes), EPITOPES):
    ax.plot([0, 1], [0, 1], ls='--', c='0.6', lw=0.8)

    for frac in FRACS:
        t = tempo(ep, frac)
        if t is None:
            continue
        d = scored(t)
        y = d['pos'].astype(int)
        fpr, tpr, _ = roc_curve(y, d['tempo'])
        a = roc_auc_score(y, d['tempo'])
        ax.plot(fpr, tpr, lw=1.3, color=pal[frac], label=f'TEMPO {frac} ({a:.3f})')
        res.append(dict(peptide=ep, score=f'TEMPO {frac}', n=len(d), auc=a,
                        auc01=roc_auc_score(y, d['tempo'], max_fpr=0.1)))

    t = tempo(ep, ALL_FRAC)                     # no-selection control
    if t is not None:
        d = scored(t)
        y = d['pos'].astype(int)
        fpr, tpr, _ = roc_curve(y, d['tempo'])
        a = roc_auc_score(y, d['tempo'])
        ax.plot(fpr, tpr, lw=1.6, color='#9e9e9e', ls=(0, (1, 1)),
                label=f'TEMPO all ({a:.3f})')
        res.append(dict(peptide=ep, score=ALL_LABEL, n=len(d), auc=a,
                        auc01=roc_auc_score(y, d['tempo'], max_fpr=0.1)))

    rd = reference(ep)
    y = rd['pos'].astype(int)
    fpr, tpr, _ = roc_curve(y, rd['iptm'])
    a = roc_auc_score(y, rd['iptm'])
    ax.plot(fpr, tpr, lw=1.6, color='black', ls=(0, (4, 2)), label=f'iptm ({a:.3f})')
    res.append(dict(peptide=ep, score=REF_LABEL, n=len(rd), auc=a,
                    auc01=roc_auc_score(y, rd['iptm'], max_fpr=0.1)))

    # combine: rank-normalise both within the epitope, then blend
    tc = tempo(ep, COMBO_FRAC)
    if tc is not None:
        c = rd[['iptm', 'pos']].join(tc, how='inner').dropna()
        yc = c['pos'].astype(int)
        for w in np.round(np.arange(0, 1.01, 0.1), 2):
            v = w * R(c['tempo'].values) + (1 - w) * R(c['iptm'].values)
            sweep.append(dict(peptide=ep, w=w, auc=roc_auc_score(yc, v),
                              auc01=roc_auc_score(yc, v, max_fpr=0.1)))
        v = W * R(c['tempo'].values) + (1 - W) * R(c['iptm'].values)
        fpr, tpr, _ = roc_curve(yc, v)
        a = roc_auc_score(yc, v)
        ax.plot(fpr, tpr, lw=1.9, color='#D62728', label=f'combo ({a:.3f})')
        res.append(dict(peptide=ep, score=COMBO_LABEL, n=len(c), auc=a,
                        auc01=roc_auc_score(yc, v, max_fpr=0.1)))
        res.append(dict(peptide=ep, score='corr(TEMPO, iptm)', n=len(c),
                        auc=np.corrcoef(R(c['tempo']), R(c['iptm']))[0, 1], auc01=np.nan))

    ax.set_title(f'{ep}', fontsize='small')
    ax.set_xlabel('FPR'); ax.set_aspect('equal')
    ax.legend(frameon=False, fontsize=5.5, loc='lower right', handlelength=2.6)

np.atleast_1d(axes)[0].set_ylabel('TPR')
fig.suptitle('TEMPO, the structural score, and their rank combination', fontsize='small')
sns.despine(); plt.tight_layout(); plt.show()

r = pd.DataFrame(res)
order = ([f'TEMPO {f}' for f in FRACS] + [ALL_LABEL, REF_LABEL, COMBO_LABEL,
                                          'corr(TEMPO, iptm)'])
for m in ['auc', 'auc01']:
    t = r.pivot(index='score', columns='peptide', values=m)[EPITOPES]
    t['MEAN'] = t.mean(axis=1)
    print(f'\n{m.upper()}')
    display(t.reindex(order).round(3))

s = pd.DataFrame(sweep).pivot(index='w', columns='peptide', values='auc01')[EPITOPES]
s['MEAN'] = s.mean(axis=1)
print('\nweight sweep, AUC01   (w=0 is pure iptm, w=1 pure TEMPO)')
display(s.round(3))
```

```python
import os
import pickle

import numpy as np
import pandas as pd
from Bio.PDB import MMCIFParser, PDBParser, Superimposer
from Bio.SeqUtils import seq1

RUNS = {'Fast-Opt': 'ESMFold2_predictions_clariden_optimized',
        'Full-Opt': 'ESMFold2_predictions_clariden_full_checkpoint_optimized'}
SCORE = 'iptm_pair_mean'
PAPER = '/Users/roessner/Documents/PostDoc/Data/260922_paper_YF_model'
GENE_REF = '/Users/roessner/Documents/PostDoc/Data/MixTCRviz/data_raw/CDR123/HomoSapiens'
EXP_REF_PDB = f'{PAPER}/exp_structures/pdbs_mhc_align/1AO7.pdb'
CACHE = 'raw_cdr_traces_bestfold.pkl'
N_MHC_CA = 180
NAMES = list(RUNS)
REF = RUNS[NAMES[0]]

ungap = lambda s: str(s).replace('-', '').replace('.', '').upper()
V_REF = {'TRA': pd.read_csv(f'{GENE_REF}/TRAV.csv', index_col=0),
         'TRB': pd.read_csv(f'{GENE_REF}/TRBV.csv', index_col=0)}
ref = [r['CA'] for r in PDBParser(QUIET=True).get_structure('r', EXP_REF_PDB)[0]['A']
       if r.id[0] == ' ' and 'CA' in r][:N_MHC_CA]
parser = MMCIFParser(QUIET=True)


def traces(cif, cdrs):
    """superpose on MHC, return the raw CA trace of each of the six CDRs"""
    st = parser.get_structure('m', cif)[0]
    mob = [r['CA'] for r in st['MHC'] if r.id[0] == ' ' and 'CA' in r][:N_MHC_CA]
    if len(mob) != N_MHC_CA:
        return None, 'MHC too short'
    sup = Superimposer()
    sup.set_atoms(ref, mob)
    sup.apply(list(st.get_atoms()))
    out = {}
    for arm, chain in [('TRA', 'TCRA'), ('TRB', 'TCRB')]:
        res = [r for r in st[chain] if r.id[0] == ' ' and 'CA' in r]
        seq = ''.join(seq1(r.get_resname()) for r in res)
        out[arm] = {}
        for i, cd in enumerate(cdrs[arm], 1):
            j = seq.find(cd)
            if j < 0:
                return None, f'{arm} cdr{i} not found'
            out[arm][f'cdr{i}'] = np.array([r['CA'].get_coord()
                                            for r in res[j:j + len(cd)]])
    return out, sup.rms


avail = {k: {p for p in os.listdir(r) if os.path.exists(f'{r}/{p}/output.csv')}
         for k, r in RUNS.items()}
peptides = sorted(set.intersection(*avail.values()))

mod_traces, mod_meta, skipped = {}, [], []
for pep in peptides:
    # pool every fold from both runs, ranked so the best per TCR comes first
    folds = []
    for run, root in RUNS.items():
        o = pd.read_csv(f'{root}/{pep}/output.csv')
        o['run'], o['root'] = run, root
        folds.append(o)
    folds = pd.concat(folds, ignore_index=True).sort_values(SCORE, ascending=False)

    tbl = pd.read_csv(f'{REF}/{pep}/{pep}.csv').set_index('id')
    n_ok = 0
    for tid, r in tbl.iterrows():
        cand = folds[folds['ID'] == tid]
        if cand.empty:
            skipped.append((pep, tid, 'no fold')); continue
        try:
            cdrs = {'TRA': [ungap(V_REF['TRA'].loc[r.TRAV, 'CDR1']),
                            ungap(V_REF['TRA'].loc[r.TRAV, 'CDR2']), r.cdr3_TRA],
                    'TRB': [ungap(V_REF['TRB'].loc[r.TRBV, 'CDR1']),
                            ungap(V_REF['TRB'].loc[r.TRBV, 'CDR2']), r.cdr3_TRB]}
        except KeyError as e:
            skipped.append((pep, tid, f'gene {e}')); continue

        # highest-scoring fold across BOTH runs and all seeds that is on disk and parses
        got = None
        for _, c in cand.iterrows():
            cif = f"{c['root']}/{pep}/models/{tid}_seed{int(c.seed)}_sample{int(c.sample_idx)}.cif"
            if not os.path.exists(cif):
                continue
            t, info = traces(cif, cdrs)
            if t is not None:
                got = (t, info, c)
                break
        if got is None:
            skipped.append((pep, tid, 'no usable cif')); continue

        t, rms, c = got
        mod_traces[(pep, tid)] = t
        mod_meta.append({'peptide': pep, 'id': tid, 'tcr': r.tcr,
                         'run': c['run'], 'seed': int(c.seed),
                         'sample_idx': int(c.sample_idx), 'score': c[SCORE],
                         'mhc_rmsd': rms})
        n_ok += 1
    print(f'{pep}: {n_ok}/{len(tbl)}')

mm = pd.DataFrame(mod_meta)
with open(CACHE, 'wb') as fh:
    pickle.dump({'mod_traces': mod_traces, 'mod_meta': mm}, fh)

print(f'\n{len(mod_traces)} traces -> {CACHE}')
print('MHC RMSD median %.2f  max %.2f' % (mm.mhc_rmsd.median(), mm.mhc_rmsd.max()))
if skipped:
    display(pd.DataFrame(skipped, columns=['peptide', 'id', 'why'])['why'].value_counts())
display(mm.groupby('peptide')['run'].value_counts(normalize=True).unstack().round(2))
```

```python
import pickle

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.lines import Line2D
from sklearn.decomposition import PCA

PRED_ROOT = '.'
CACHE = 'raw_cdr_traces_bestfold.pkl'
MIN_SCORE = 0.6            # see note below: 0.0 is unusable here, 0.8 guts two epitopes
DROP_AMBIGUOUS = True
NCOL = 4
CDRS = ['cdr1', 'cdr2', 'cdr3']
COLORS = {'binder': '#1f77b4', 'decoy': '#ff7f0e', 'unlabelled': '#9e9e9e'}

with open(CACHE, 'rb') as fh:
    D = pickle.load(fh)
mod_traces = D['mod_traces']
mod_meta = D['mod_meta'].set_index(['peptide', 'id'])

meta = pd.read_csv('ttd_gil_ylq_nlv_glc_01-10-2026_TCRvdb.csv')
lab = meta.set_index('tcr_id')[['pos', 'ambiguous']]


def resample_loop(coords, n):
    coords = np.asarray(coords, float)
    if len(coords) == n:
        return coords
    if len(coords) == 1:
        return np.repeat(coords, n, axis=0)
    d = np.r_[0, np.cumsum(np.linalg.norm(np.diff(coords, axis=0), axis=1))]
    if d[-1] == 0:
        return np.repeat(coords[:1], n, axis=0)
    u = d / d[-1]
    return np.column_stack([np.interp(np.linspace(0, 1, n), u, coords[:, k])
                            for k in range(3)])


def vectorise(keys, targets):
    X = []
    for k in keys:
        t, vec = mod_traces[k], []
        for arm in ['TRA', 'TRB']:
            for c in CDRS:
                vec.extend(resample_loop(t[arm][c], targets[arm][c]).flatten())
        X.append(vec)
    return np.asarray(X)


peptides = sorted({p for p, _ in mod_traces})
nrow = int(np.ceil(len(peptides) / NCOL))
fig, axes = plt.subplots(nrow, NCOL, figsize=(3.1 * NCOL, 3.0 * nrow))
axes = np.atleast_1d(axes).ravel()
report = []

for ax, peptide in zip(axes, peptides):
    keys = [k for k in mod_traces if k[0] == peptide
            and mod_meta.loc[k, 'score'] >= MIN_SCORE]

    info = mod_meta.loc[keys].reset_index().join(lab, on='tcr')
    if info['pos'].notna().any():
        if DROP_AMBIGUOUS:
            keep = (info['ambiguous'] != 1).to_numpy()
            info = info[keep].reset_index(drop=True)
            keys = [k for k, m in zip(keys, keep) if m]
        info['label'] = info['pos'].map({1: 'binder', 0: 'decoy'})
    else:
        info['label'] = 'unlabelled'

    # points per loop = median loop length across this epitope's retained models
    targets = {arm: {c: int(np.round(np.median([len(mod_traces[k][arm][c]) for k in keys])))
                     for c in CDRS} for arm in ['TRA', 'TRB']}

    X = vectorise(keys, targets)
    pca = PCA(n_components=2, random_state=0).fit(X)       # fit on the MODELS themselves
    pcs = pca.transform(X)

    for lb in ['decoy', 'binder', 'unlabelled']:
        m = (info['label'] == lb).to_numpy()
        if m.any():
            ax.scatter(pcs[m, 0], pcs[m, 1], s=7, color=COLORS[lb], alpha=0.55, linewidths=0)

    lo, hi = np.quantile(pcs, [0.005, 0.995], axis=0)
    ax.set_xlim(lo[0] - 8, hi[0] + 8)
    ax.set_ylim(lo[1] - 8, hi[1] + 8)

    t = targets
    ax.set_title(f"{peptide}  (n={len(keys)})\n"
                 f"a {t['TRA']['cdr1']}/{t['TRA']['cdr2']}/{t['TRA']['cdr3']}  "
                 f"b {t['TRB']['cdr1']}/{t['TRB']['cdr2']}/{t['TRB']['cdr3']}",
                 fontsize='small')
    ax.set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]*100:.0f}%)')
    ax.set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]*100:.0f}%)')
    report.append(dict(peptide=peptide, n=len(keys), n_features=X.shape[1],
                    **info['label'].value_counts().to_dict(),
                    **{f'from_{k}': v for k, v in
                        info['run'].value_counts(normalize=True).round(2).items()},
                    var_PC1=pca.explained_variance_ratio_[0],
                    var_PC2=pca.explained_variance_ratio_[1]))

for ax in axes[len(peptides):]:
    ax.axis('off')

axes[0].legend(handles=[Line2D([0], [0], marker='o', ls='', ms=4, color=c, label=k)
                        for k, c in COLORS.items()],
               frameon=False, loc='upper left', fontsize=7,
               handlelength=1, handletextpad=0.5)

sns.despine()
plt.tight_layout()
plt.show()

display(pd.DataFrame(report))
```

```python
import os
import pickle
import itertools

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import rankdata
from scipy.spatial.distance import squareform, pdist
from sklearn.metrics import roc_auc_score, roc_curve

EPITOPES = ['TTDPSFLGRY', 'GILGFVFTL', 'YLQPRTFLL', 'NLVPMVATV', 'GLCTLVAML']
TEMPO_FRAC = '1.0'
OUT_ROOT = 'TEMPO_out'
DROP_AMBIGUOUS = True
TRACES = 'raw_cdr_traces_bestfold.pkl'
CDRS = ['cdr1', 'cdr2', 'cdr3']
K_NN = 3                      # 3 neighbours: the sweep optimum

REF_RUNS = {'Fast-Opt': 'ESMFold2_predictions_clariden_optimized',
            'Full-Opt': 'ESMFold2_predictions_clariden_full_checkpoint_optimized'}
REF_SCORE = 'iptm_pair_mean'

STYLE = {'TEMPO': ('#55A868', '-', 1.5),
         'iptm': ('black', (0, (4, 2)), 1.6),
         'conformational density': ('#8172B3', '-', 1.5),
         'combo TEMPO+iptm': ('#D62728', '-', 1.8),
         'combo all three': ('#C44E52', (0, (1, 1)), 2.2)}

meta = pd.read_csv('ttd_gil_ylq_nlv_glc_01-10-2026_TCRvdb.csv')
labels = meta.set_index('tcr_id')[['pos', 'ambiguous']]
R = lambda v: rankdata(v) / len(v)          # rank-normalise: the scales differ

with open(TRACES, 'rb') as fh:
    D = pickle.load(fh)
mod_traces, mod_meta = D['mod_traces'], D['mod_meta'].set_index(['peptide', 'id'])


def resample_loop(c, n):
    """treat a CA trace as a 3D curve, resample to n points evenly along arc length"""
    c = np.asarray(c, float)
    if len(c) == n:
        return c
    d = np.r_[0, np.cumsum(np.linalg.norm(np.diff(c, axis=0), axis=1))]
    if d[-1] == 0:
        return np.repeat(c[:1], n, axis=0)
    u = d / d[-1]
    return np.column_stack([np.interp(np.linspace(0, 1, n), u, c[:, k]) for k in range(3)])


def density(ep):
    """label-free: -mean distance to the K_NN nearest conformational neighbours"""
    keys = [k for k in mod_traces if k[0] == ep]
    tcr = mod_meta.loc[keys, 'tcr'].to_numpy()
    targets = {a: {c: int(np.round(np.median([len(mod_traces[k][a][c]) for k in keys])))
                   for c in CDRS} for a in ['TRA', 'TRB']}
    X = np.array([[x for a in ['TRA', 'TRB'] for c in CDRS
                   for x in resample_loop(mod_traces[k][a][c], targets[a][c]).flatten()]
                  for k in keys])
    M = squareform(pdist(X))          # raw coordinates: PCA here would be an exact no-op
    np.fill_diagonal(M, np.inf)
    return pd.Series(-np.sort(M, axis=1)[:, :K_NN].mean(1), index=tcr, name='dens')


def iptm(ep):
    cols = {}
    for run, root in REF_RUNS.items():
        sc = (pd.read_csv(f'{root}/{ep}/output.csv')
              .groupby('ID', sort=False)[REF_SCORE].max().rename(run))
        tbl = pd.read_csv(f'{root}/{ep}/{ep}.csv').set_index('id')
        cols[run] = tbl.join(sc, how='inner').set_index('tcr')[run]
    return pd.concat(cols.values(), axis=1, join='inner').max(axis=1).rename('iptm')


def tempo(ep):
    p = f'{OUT_ROOT}/{ep}_{TEMPO_FRAC}/all_tcrs.csv'
    return None if not os.path.exists(p) else \
        pd.read_csv(p).set_index('tcr')['score'].rename('tempo')


res, corrs = [], []
fig, axes = plt.subplots(1, len(EPITOPES), figsize=(3.3 * len(EPITOPES), 3.6),
                         sharex=True, sharey=True)
for ax, ep in zip(np.atleast_1d(axes), EPITOPES):
    d = pd.concat([iptm(ep), tempo(ep), density(ep)], axis=1, join='inner').join(labels)
    if DROP_AMBIGUOUS:
        d = d[d['ambiguous'] == 0]
    d = d.dropna(subset=['iptm', 'tempo', 'dens', 'pos'])
    y = d['pos'].astype(int)

    rt, ri, rd = R(d.tempo.values), R(d.iptm.values), R(d.dens.values)
    scores = {'TEMPO': d.tempo.values, 'iptm': d.iptm.values,
              'conformational density': d.dens.values,
              'combo TEMPO+iptm': (rt + ri) / 2,
              'combo all three': (rt + ri + rd) / 3}

    ax.plot([0, 1], [0, 1], ls=':', c='0.6', lw=0.8)
    for name, v in scores.items():
        c, ls, lw = STYLE[name]
        fpr, tpr, _ = roc_curve(y, v)
        a = roc_auc_score(y, v)
        ax.plot(fpr, tpr, lw=lw, color=c, ls=ls, label=f'{name} ({a:.3f})')
        res.append(dict(peptide=ep, score=name, n=len(d), auc=a,
                        auc01=roc_auc_score(y, v, max_fpr=0.1)))

    for a_, b_ in itertools.combinations(['tempo', 'iptm', 'dens'], 2):
        corrs.append(dict(peptide=ep, pair=f'{a_} vs {b_}',
                          r=np.corrcoef(R(d[a_].values), R(d[b_].values))[0, 1]))

    ax.set_title(f'{ep} (n={len(d)})', fontsize='small')
    ax.set_xlabel('FPR'); ax.set_aspect('equal')
    ax.legend(frameon=False, fontsize=5.8, loc='lower right', handlelength=2.6)

np.atleast_1d(axes)[0].set_ylabel('TPR')
fig.suptitle(f'TEMPO {TEMPO_FRAC}, structural confidence, conformational density, and combinations',
             fontsize='small')
sns.despine(); plt.tight_layout(); plt.show()

r = pd.DataFrame(res)
for m in ['auc', 'auc01']:
    t = r.pivot(index='score', columns='peptide', values=m)[EPITOPES]
    t['MEAN'] = t.mean(axis=1)
    print(f'\n{m.upper()}')
    display(t.reindex(list(STYLE)).round(3))

print('\nrank correlations between the three components')
display(pd.DataFrame(corrs).pivot(index='peptide', columns='pair', values='r')
        .loc[EPITOPES].round(3))
```

```python
import os
import pickle

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import rankdata
from scipy.spatial.distance import squareform, pdist
from sklearn.metrics import roc_auc_score, roc_curve
from tcrdist.repertoire import TCRrep

REF_RUNS = {'Fast-Opt': 'ESMFold2_predictions_clariden_optimized',
            'Full-Opt': 'ESMFold2_predictions_clariden_full_checkpoint_optimized'}
ROOT = REF_RUNS['Fast-Opt']
TRACES = 'raw_cdr_traces_bestfold.pkl'
PW_CACHE = 'tcrdist_pw.pkl'
EPITOPES = ['TTDPSFLGRY', 'GILGFVFTL', 'YLQPRTFLL', 'NLVPMVATV', 'GLCTLVAML']
CDRS = ['cdr1', 'cdr2', 'cdr3']
K_NN = 3
TEMPO_FRAC = 'all'
ALPHA = 0.5                       # weight on SEQUENCE in the early fusion
R = lambda v: rankdata(v) / len(v)

STYLE = {'sequence density': ('#DD8452', (0, (4, 2)), 1.3),
         'structure density': ('#8172B3', (0, (4, 2)), 1.3),
         'iptm': ('black', (0, (4, 2)), 1.3),
         'TEMPO (all, reference)': ('#CCB974', (0, (6, 2, 1, 2)), 1.5),
         'seq + struct (late fusion)': ('#4C72B0', '-', 1.6),
         'fused distance (early fusion)': ('#C44E52', '-', 1.6),
         'seq + iptm': ('#55A868', '-', 1.3),
         'struct + iptm': ('#937860', '-', 1.3),
         'seq + struct + iptm': ('#C44E52', (0, (1, 1)), 2.4),
         'fused distance + iptm': ('#4C72B0', (0, (1, 1)), 2.4)}

meta = pd.read_csv('ttd_gil_ylq_nlv_glc_01-10-2026_TCRvdb.csv')
labels = meta.set_index('tcr_id')[['pos', 'ambiguous']]
D = pickle.load(open(TRACES, 'rb'))
mt, mm = D['mod_traces'], D['mod_meta'].set_index(['peptide', 'id'])


def resample_loop(c, n):
    c = np.asarray(c, float)
    if len(c) == n:
        return c
    d = np.r_[0, np.cumsum(np.linalg.norm(np.diff(c, axis=0), axis=1))]
    if d[-1] == 0:
        return np.repeat(c[:1], n, axis=0)
    u = d / d[-1]
    return np.column_stack([np.interp(np.linspace(0, 1, n), u, c[:, k]) for k in range(3)])


def tcrdist_pw(peps):
    """pairwise tcrdist per epitope, cached (needs the wheeler kernel)"""
    if os.path.exists(PW_CACHE):
        return pickle.load(open(PW_CACHE, 'rb'))
    PW = {}
    for p in peps:
        d = pd.read_csv(f'{ROOT}/{p}/{p}.csv')
        df = pd.DataFrame({'cdr3_a_aa': d.cdr3_TRA, 'v_a_gene': d.TRAV + '*01',
                           'j_a_gene': d.TRAJ + '*01', 'cdr3_b_aa': d.cdr3_TRB,
                           'v_b_gene': d.TRBV + '*01', 'j_b_gene': d.TRBJ + '*01',
                           'count': 1, 'tcr': d.tcr})
        tr = TCRrep(cell_df=df, organism='human', chains=['alpha', 'beta'],
                    infer_all_genes=True, compute_distances=True, deduplicate=False)
        PW[p] = {'tcr': tr.clone_df['tcr'].to_numpy(),
                 'D': (tr.pw_alpha + tr.pw_beta).astype(float)}
        print(f'{p}: tcrdist {PW[p]["D"].shape}')
    pickle.dump(PW, open(PW_CACHE, 'wb'))
    return PW


def struct_pw(ep):
    keys = [k for k in mt if k[0] == ep]
    t = {a: {c: int(np.round(np.median([len(mt[k][a][c]) for k in keys]))) for c in CDRS}
         for a in ['TRA', 'TRB']}
    X = np.array([[x for a in ['TRA', 'TRB'] for c in CDRS
                   for x in resample_loop(mt[k][a][c], t[a][c]).flatten()] for k in keys])
    return mm.loc[keys, 'tcr'].to_numpy(), squareform(pdist(X))


def iptm(ep):
    """best iptm_pair_mean over seeds, then over both optimized checkpoints"""
    cols = {}
    for run, root in REF_RUNS.items():
        sc = (pd.read_csv(f'{root}/{ep}/output.csv')
              .groupby('ID', sort=False)['iptm_pair_mean'].max().rename(run))
        tbl = pd.read_csv(f'{root}/{ep}/{ep}.csv').set_index('id')
        cols[run] = tbl.join(sc, how='inner').set_index('tcr')[run]
    return pd.concat(cols.values(), axis=1, join='inner').max(axis=1).rename('iptm')


def tempo(ep):
    """reference only: trained on every TCR of the epitope and scored on the same set"""
    p = f'TEMPO_out/{ep}_{TEMPO_FRAC}/all_tcrs.csv'
    if not os.path.exists(p):
        return None
    return pd.read_csv(p).set_index('tcr')['score'].rename('tempo')


def rank_norm(M):
    """rank-transform off-diagonal entries to (0,1] so the two scales are comparable"""
    iu = np.triu_indices_from(M, 1)
    out = np.zeros_like(M, dtype=float)
    out[iu] = rankdata(M[iu]) / len(M[iu])
    return out + out.T


def density(M, k=K_NN):
    M = M.copy()
    np.fill_diagonal(M, np.inf)
    return -np.sort(M, axis=1)[:, :k].mean(1)


PW = tcrdist_pw(EPITOPES)
res = []
fig, axes = plt.subplots(1, len(EPITOPES), figsize=(3.3 * len(EPITOPES), 3.6),
                         sharex=True, sharey=True)
for ax, ep in zip(np.atleast_1d(axes), EPITOPES):
    tcr_s, Ds = struct_pw(ep)
    tcr_t, Dt = PW[ep]['tcr'], PW[ep]['D']
    ip = iptm(ep)

    # reorder everything onto the TCRs present in all three sources
    pos_s = {t: i for i, t in enumerate(tcr_s)}
    pos_t = {t: i for i, t in enumerate(tcr_t)}
    ok = set(pos_t) & set(ip.index)
    common = [t for t in tcr_s if t in ok]
    si, ti = [pos_s[t] for t in common], [pos_t[t] for t in common]
    Ds, Dt = Ds[np.ix_(si, si)], Dt[np.ix_(ti, ti)]

    lab = labels.reindex(common)
    keep = ((lab['ambiguous'] == 0) & lab['pos'].notna()).to_numpy()
    Ds, Dt = Ds[np.ix_(keep, keep)], Dt[np.ix_(keep, keep)]
    y = lab['pos'].to_numpy()[keep].astype(int)
    v_ip = ip.reindex(common).to_numpy()[keep]

    # early fusion blends the distances, late fusion blends the densities
    Ns, Nt = rank_norm(Ds), rank_norm(Dt)
    d_seq, d_str = density(Nt), density(Ns)
    d_fused = density(ALPHA * Nt + (1 - ALPHA) * Ns)
    rq, rt, ri, rf = R(d_seq), R(d_str), R(v_ip), R(d_fused)

    scores = {'sequence density': d_seq,
              'structure density': d_str,
              'iptm': v_ip,
              'seq + struct (late fusion)': (rq + rt) / 2,
              'fused distance (early fusion)': d_fused,
              'seq + iptm': (rq + ri) / 2,
              'struct + iptm': (rt + ri) / 2,
              'seq + struct + iptm': (rq + rt + ri) / 3,
              'fused distance + iptm': (rf + ri) / 2}

    # reference curve only, deliberately kept out of every rank combination
    tp = tempo(ep)
    if tp is not None:
        scores['TEMPO (all, reference)'] = tp.reindex(common).to_numpy()[keep]

    ax.plot([0, 1], [0, 1], ls=':', c='0.6', lw=0.8)
    for name, v in scores.items():
        c, ls, lw = STYLE[name]
        m = np.isfinite(v)                 # TEMPO may not cover every TCR
        yv, vv = y[m], v[m]
        fpr, tpr, _ = roc_curve(yv, vv)
        a = roc_auc_score(yv, vv)
        ax.plot(fpr, tpr, lw=lw, color=c, ls=ls, label=f'{name} ({a:.3f})')
        res.append(dict(peptide=ep, score=name, n=int(m.sum()), auc=a,
                        auc01=roc_auc_score(yv, vv, max_fpr=0.1)))
    ax.set_title(f'{ep} (n={len(y)})', fontsize='small')
    ax.set_xlabel('FPR'); ax.set_aspect('equal')
    ax.legend(frameon=False, fontsize=5.0, loc='lower right', handlelength=2.6)

np.atleast_1d(axes)[0].set_ylabel('TPR')
fig.suptitle('Sequence density, structure density and iptm: early vs late fusion',
             fontsize='small')
sns.despine(); plt.tight_layout(); plt.show()

r = pd.DataFrame(res)
for m in ['auc', 'auc01']:
    t = r.pivot(index='score', columns='peptide', values=m)[EPITOPES]
    t['MEAN'] = t.mean(axis=1)
    print(f'\n{m.upper()}')
    display(t.reindex(list(STYLE)).round(3))
```

```python
import os
import pickle

import numpy as np
import pandas as pd
from scipy.stats import rankdata
from scipy.spatial.distance import squareform, pdist
from sklearn.metrics import roc_auc_score
from tcrdist.repertoire import TCRrep

REF_RUNS = {'Fast-Opt': 'ESMFold2_predictions_clariden_optimized',
            'Full-Opt': 'ESMFold2_predictions_clariden_full_checkpoint_optimized'}
ROOT = REF_RUNS['Fast-Opt']
TRACES = 'raw_cdr_traces_bestfold.pkl'
PW_CACHE = 'tcrdist_pw.pkl'
TEST = 'test.csv'
OUT_CSV = 'submission_seq_struct_iptm.csv'
CDRS = ['cdr1', 'cdr2', 'cdr3']
K_NN = 3
R = lambda v: rankdata(v) / len(v)

D = pickle.load(open(TRACES, 'rb'))
mt, mm = D['mod_traces'], D['mod_meta'].set_index(['peptide', 'id'])
EPITOPES = sorted({k[0] for k in mt})

test = pd.read_csv(TEST)
labels = (pd.read_csv('ttd_gil_ylq_nlv_glc_01-10-2026_TCRvdb.csv')
          .set_index('tcr_id')[['pos', 'ambiguous']])


def resample_loop(c, n):
    c = np.asarray(c, float)
    if len(c) == n:
        return c
    d = np.r_[0, np.cumsum(np.linalg.norm(np.diff(c, axis=0), axis=1))]
    if d[-1] == 0:
        return np.repeat(c[:1], n, axis=0)
    u = d / d[-1]
    return np.column_stack([np.interp(np.linspace(0, 1, n), u, c[:, k]) for k in range(3)])


def tcrdist_pw(peps):
    """pairwise tcrdist per epitope, cached; extends the cache for new epitopes"""
    PW = pickle.load(open(PW_CACHE, 'rb')) if os.path.exists(PW_CACHE) else {}
    missing = [p for p in peps if p not in PW]
    for p in missing:
        d = pd.read_csv(f'{ROOT}/{p}/{p}.csv')
        df = pd.DataFrame({'cdr3_a_aa': d.cdr3_TRA, 'v_a_gene': d.TRAV + '*01',
                           'j_a_gene': d.TRAJ + '*01', 'cdr3_b_aa': d.cdr3_TRB,
                           'v_b_gene': d.TRBV + '*01', 'j_b_gene': d.TRBJ + '*01',
                           'count': 1, 'tcr': d.tcr})
        tr = TCRrep(cell_df=df, organism='human', chains=['alpha', 'beta'],
                    infer_all_genes=True, compute_distances=True, deduplicate=False)
        PW[p] = {'tcr': tr.clone_df['tcr'].to_numpy(),
                 'D': (tr.pw_alpha + tr.pw_beta).astype(float)}
        print(f'  tcrdist {p}: {PW[p]["D"].shape}')
    if missing:
        pickle.dump(PW, open(PW_CACHE, 'wb'))
    return PW


def struct_pw(ep):
    keys = [k for k in mt if k[0] == ep]
    t = {a: {c: int(np.round(np.median([len(mt[k][a][c]) for k in keys]))) for c in CDRS}
         for a in ['TRA', 'TRB']}
    X = np.array([[x for a in ['TRA', 'TRB'] for c in CDRS
                   for x in resample_loop(mt[k][a][c], t[a][c]).flatten()] for k in keys])
    return mm.loc[keys, 'tcr'].to_numpy(), squareform(pdist(X))


def iptm(ep):
    """best iptm_pair_mean over seeds, then over both optimized checkpoints"""
    cols = {}
    for run, root in REF_RUNS.items():
        sc = (pd.read_csv(f'{root}/{ep}/output.csv')
              .groupby('ID', sort=False)['iptm_pair_mean'].max().rename(run))
        tbl = pd.read_csv(f'{root}/{ep}/{ep}.csv').set_index('id')
        cols[run] = tbl.join(sc, how='inner').set_index('tcr')[run]
    return pd.concat(cols.values(), axis=1, join='inner').max(axis=1).rename('iptm')


def rank_norm(M):
    """rank-transform off-diagonal entries to (0,1] so the two scales are comparable"""
    iu = np.triu_indices_from(M, 1)
    out = np.zeros_like(M, dtype=float)
    out[iu] = rankdata(M[iu]) / len(M[iu])
    return out + out.T


def density(M, k=K_NN):
    M = M.copy()
    np.fill_diagonal(M, np.inf)
    return -np.sort(M, axis=1)[:, :k].mean(1)


PW = tcrdist_pw(EPITOPES)

out, audit = [], []
for ep in EPITOPES:
    tcr_s, Ds = struct_pw(ep)
    tcr_t, Dt = PW[ep]['tcr'], PW[ep]['D']
    ip = iptm(ep)

    # reorder everything onto the TCRs present in all three sources
    pos_s = {t: i for i, t in enumerate(tcr_s)}
    pos_t = {t: i for i, t in enumerate(tcr_t)}
    ok = set(pos_t) & set(ip.index)
    common = [t for t in tcr_s if t in ok]
    si, ti = [pos_s[t] for t in common], [pos_t[t] for t in common]
    Ds, Dt = Ds[np.ix_(si, si)], Dt[np.ix_(ti, ti)]
    v_ip = ip.reindex(common).to_numpy()

    # neighbourhoods use every TCR of the epitope, labelled or not
    pred = (R(density(rank_norm(Dt))) + R(density(rank_norm(Ds))) + R(v_ip)) / 3
    out.append(pd.DataFrame({'tcr_id': common, 'prediction': pred, 'epitope': ep}))

    row = dict(epitope=ep, n_test=int(test.tcr_id.str.endswith('_' + ep).sum()),
               n_traces=len(tcr_s), n_scored=len(common))
    lab = labels.reindex(common)
    k = ((lab['ambiguous'] == 0) & lab['pos'].notna()).to_numpy()
    if k.sum() and 0 < lab['pos'].to_numpy()[k].sum() < k.sum():
        row.update(auc=roc_auc_score(lab['pos'].to_numpy()[k].astype(int), pred[k]),
                   auc01=roc_auc_score(lab['pos'].to_numpy()[k].astype(int), pred[k],
                                       max_fpr=0.1),
                   n_lab=int(k.sum()))
    audit.append(row)

sub = pd.concat(out, ignore_index=True)

a = pd.DataFrame(audit)
display(a.round(3))
test_ids, got = set(test.tcr_id), set(sub.tcr_id)
print(f'test.csv {len(test_ids)}   scored {len(got)}   '
      f'missing {len(test_ids - got)}   extra {len(got - test_ids)}   '
      f'duplicated {sub.tcr_id.duplicated().sum()}')
print(f'mean over labelled: auc={a.auc.mean():.3f}  auc01={a.auc01.mean():.3f}')

# epitope blocks, best (highest score) first within each block
sub['epitope'] = pd.Categorical(sub['epitope'], categories=EPITOPES, ordered=True)
sub = (sub.sort_values(['epitope', 'prediction'], ascending=[True, False])
          .reset_index(drop=True))
sub['epitope'] = sub['epitope'].astype(str)
sub.to_csv(OUT_CSV, index=False)
print(f'wrote {OUT_CSV}  rows={len(sub)}  nan={int(sub.prediction.isna().sum())}')
display(sub.groupby('epitope', sort=False).head(2))
print('monotone within every block:',
      all(g.prediction.is_monotonic_decreasing for _, g in sub.groupby('epitope')))
```
